```python
#binary search
a = list(map(int, input("Enter numbers: ").split()))
tar = int(input("Enter target number: "))

low = 0
high = len(a) - 1

while low <= high:
    mid = (low + high) // 2

    if a[mid] == tar:
        print("Found at index:", mid)
        break

    elif a[mid] < tar:
        low = mid + 1

    else:
        high = mid - 1

else:
    print("Target not found")
```

```python
#selection sort
a = list(map(int, input("Enter numbers: ").split()))

n = len(a)

for i in range(n):
    min = i

    for j in range(i + 1, n):
        if a[j] < a[min]:
            min = j

    a[i], a[min] = a[min], a[i]

print("Sorted array:", a)
```

```python
#insertion sort
a = list(map(int, input("Enter numbers: ").split()))

for i in range(1, len(a)):
    key = a[i]
    j = i - 1

    while j >= 0 and a[j] > key:
        a[j + 1] = a[j]
        j -= 1

    a[j + 1] = key

print("Sorted array:", a)
```

```python
#bubble sort
a = list(map(int, input("Enter numbers: 1,2,3,4").split()))

n = len(a)

for i in range(n):
    for j in range(0, n - i - 1):

        if a[j] > a[j + 1]:
            a[j], a[j + 1] = a[j + 1], a[j]

print("Sorted array:", a)
```

```python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2

        left = arr[:mid]
        right = arr[mid:]

        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        # Merge the two halves
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        # Copy remaining elements of left
        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        # Copy remaining elements of right
        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1


arr = list(map(int, input("Enter numbers: ").split()))

merge_sort(arr)

print("Sorted array:", arr)
```

```python
#quick sort
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1


def quick_sort(arr, low, high):
    if low < high:
        pivot_index = partition(arr, low, high)

        quick_sort(arr, low, pivot_index - 1)
        quick_sort(arr, pivot_index + 1, high)


arr = list(map(int, input("Enter numbers: ").split()))

quick_sort(arr, 0, len(arr) - 1)

print("Sorted array:", arr)
```

```python
#heap-mini
def heapify(arr, n, i):
    smallest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] < arr[smallest]:
        smallest = left

    if right < n and arr[right] < arr[smallest]:
        smallest = right

    if smallest != i:
        arr[i], arr[smallest] = arr[smallest], arr[i]
        heapify(arr, n, smallest)


def min_heap_sort(arr):
    n = len(arr)


    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    result = []

    for i in range(n):
        result.append(arr[0])

        arr[0], arr[n - i - 1] = arr[n - i - 1], arr[0]

        heapify(arr, n - i - 1, 0)

    return result


arr = list(map(int, input("Enter elements: ").split()))

print("Sorted array:", min_heap_sort(arr))
```

```python
#heap-max
def heapify(arr, n, i):
    largest = i

    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left

    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]

        heapify(arr, n, largest)


def max_heap_sort(arr):
    n = len(arr)


    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)
    for i in range(n - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]

        heapify(arr, i, 0)


arr = list(map(int, input("Enter elements: ").split()))

max_heap_sort(arr)

print("Sorted array:", arr)
```

```python
#knapsack
def knapsack(weights, values, capacity):
    n = len(weights)

    dp = [[0] * (capacity + 1) for i in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(1, capacity + 1):

            if weights[i - 1] <= w:
                dp[i][w] = max(
                    values[i - 1] + dp[ i- 1][w - weights[i - 1]],
                    dp[i - 1][w]
                )
            else:
                dp[i][w] = dp[i - 1][w]

    return dp[n][capacity]


weights = list(map(int, input("Enter weights: ").split()))
values = list(map(int, input("Enter values: ").split()))
capacity = int(input("Enter capacity: "))

print("Maximum value:", knapsack(weights, values, capacity))
```

```python
#matrix
def matrix_chain(p, n):
    dp = [[0 for j in range(n)] for i in range(n)]

    for length in range(2, n):
        for i in range(1, n - length + 1):
            j = i + length - 1
            dp[i][j] = 999999

            for k in range(i, j):
                cost = (dp[i][k] + dp[k + 1][j]
                        + p[i - 1] * p[k] * p[j])

                if cost < dp[i][j]:
                    dp[i][j] = cost

    return dp[1][n - 1]



n = int(input("Enter number of matrices: "))

p = []

print("Enter dimensions:")
for i in range(n + 1):
    x = int(input())
    p.append(x)

ans = matrix_chain(p, n)

print("Minimum number of multiplications:", ans)
```

```python
def coin_change(coins, amount):
    n = len(coins)

    dp = [[0] * (amount + 1) for _ in range(n + 1)]

    for i in range(n + 1):
        dp[i][0] = 1

    for i in range(1, n + 1):
        for j in range(1, amount + 1):

            if coins[i - 1] <= j:
                dp[i][j] = dp[i - 1][j] + dp[i][j - coins[i - 1]]
            else:
                dp[i][j] = dp[i - 1][j]

    print("\nDP Table:")

    for row in dp:
        print(row)

    return dp[n][amount]


coins = list(map(int, input("Enter coins: ").split()))
amount = int(input("Enter amount: "))

ways = coin_change(coins, amount)

print("Number of ways:", ways)
```

```python
#bfs
from collections import deque

def bfs(adj):
    n = len(adj)
    visited = [False] * n
    res = []

    src = 0
    g = deque()

    g.append(src)
    visited[src] = True
    while g:
        u = g.popleft()
        res.append(u)

        for v in adj[u]:
            if not visited[v]:
                g.append(v)
                visited[v] = True

    return res

adj = [
    [1, 2],
    [0, 3],
    [0, 3],
    [1, 2, 4],
    [3]
]

print("BFS Traversal:", bfs(adj))
```

```python
def find_parent(parent, i):
    if parent[i] == i:
        return i
    return find_parent(parent, parent[i])


def union_set(parent, u, v):
    parent_u = find_parent(parent, u)
    parent_v = find_parent(parent, v)
    parent[parent_u] = parent_v


n = int(input("Enter number of vertices: "))
e = int(input("Enter number of edges: "))

edges = []

print("Enter edges (source destination weight):")

for i in range(e):
    u, v, w = map(int, input().split())
    edges.append((w, u, v))

# Sort edges by weight
edges.sort()

# Initialize parent
parent = [i for i in range(n)]

total_cost = 0
count = 0

print("\nEdges in Minimum Spanning Tree:")

for w, u, v in edges:

    if find_parent(parent, u) != find_parent(parent, v):

        print(u, "-", v, ":", w)

        total_cost += w
        union_set(parent, u, v)
        count += 1

    if count == n - 1:
        break

print("Minimum Cost =", total_cost)
```

```python
n = int(input("Enter number of vertices: "))

dist = []

print("Enter adjacency matrix (-1 for no edge):")

for i in range(n):
    row = list(map(int, input().split()))

    for j in range(n):
        if row[j] == -1:
            row[j] = 99999

    dist.append(row)

# Floyd-Warshall Algorithm
for k in range(n):
    for i in range(n):
        for j in range(n):

            if dist[i][k] != 99999 and dist[k][j] != 99999:
                if dist[i][k] + dist[k][j] < dist[i][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]

# Display shortest distance matrix
print("Shortest distance matrix:")

for i in range(n):
    for j in range(n):
        if dist[i][j] == 99999:
            print("INF", end=" ")
        else:
            print(dist[i][j], end=" ")
    print()
```

```python
from itertools import permutations

n = int(input("Enter number of cities: "))

cost = []

print("Enter cost matrix:")

for i in range(n):
    row = list(map(int, input().split()))
    cost.append(row)

# Generate all possible paths
path = list(range(1, n))

min_cost = float('inf')
best_path = []

for p in permutations(path):

    current_cost = cost[0][p[0]]

    for i in range(n - 2):
        current_cost += cost[p[i]][p[i + 1]]

    current_cost += cost[p[n - 2]][0]

    if current_cost < min_cost:
        min_cost = current_cost
        best_path = [0] + list(p)

print("Minimum cost:", min_cost)

print("Optimal tour:", end=" ")

for city in best_path:
    print(city, "->", end=" ")

print("0")
```
